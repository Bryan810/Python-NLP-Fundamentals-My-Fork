```python
from string import punctuation   # Importa la constante 'punctuation' del módulo 'string'
print(punctuation)               # Imprime todos los caracteres de puntuación definidos en ASCII
```

```python
def remove_punct(text):                               # Define una función que recibe un texto como entrada
    '''Remove punctuation marks in input text'''      
    
    no_punct = []                                     # Lista vacía para almacenar los caracteres válidos
    for char in text:                                 # Recorre cada carácter del texto
        if char not in punctuation:                   # Verifica si el carácter no es un signo de puntuación
            no_punct.append(char)                     # Si no lo es, lo añade a la lista
    text_no_punct = ''.join(no_punct)                 # Une la lista de caracteres en una cadena
    
    return text_no_punct                              # Devuelve el texto sin signos de puntuación
```

```python
third_example = tweets['text'][20]          # Selecciona el tweet número 20 (índice 20) de la columna 'text' del DataFrame 'tweets'
print(third_example)                        # Imprime el tweet original con toda su puntuación
print(f"{'=' * 50}")                        # Imprime una línea de 50 signos '=' como separador visual
remove_punct(third_example)                 # Llama a la función definida antes para eliminar los signos de puntuación
```

```python
challenge1_path = '../data/example1.txt'   # Define la ruta relativa al archivo de texto que se quiere abrir

with open(challenge1_path, 'r') as file:   # Abre el archivo en modo lectura ('r')
    challenge1 = file.read()               # Lee todo el contenido del archivo y lo guarda en la variable 'challenge1'
    
print(challenge1)                          # Imprime el contenido del archivo en pantalla
```

```python

# Función para limpiar el texto
blankspace_pattern = r'\s+'
blankspace_repl = ' '
def clean_text(text):
        
    # 1. Convertir a minúsculas
    text = text.lower()                                  
    # 2. Eliminar signos de puntuación
    text = remove_punct(text)                       
    # 3. Eliminar espacios en blanco adicionales
    text = re.sub(blankspace_pattern, blankspace_repl, text)
    text = text.strip()
    
    return text
    
clean_text(challenge1)
```

Remove Hashtags and URLs

```python
url_tweet = tweets['text'][13]   # Selecciona el tweet número 13 (índice 13) de la columna 'text' del DataFrame 'tweets'
print(url_tweet)                 # Imprime el contenido del tweet 
```

```python
# Expresión regular para identificar URLs que empiezan con http, https o ftp,
url_pattern = r'(http|ftp|https):\/\/([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:\/~+#-]*[\w@?^=%&\/~+#-])'

url_repl = ' URL '    # Texto que usaremos como reemplazo de cualquier URL encontrada

re.sub(url_pattern, url_repl, url_tweet)  # Busca todas las coincidencias de URLs en el tweet y las reemplaza por ' URL '
```

```python

hashtag_pattern = r'(?:^|\s)[＃#]{1}(\w+)'        # Expresión regular para detectar hashtags
hashtag_repl = ' HASHTAG '                        # Reemplazo: cada hashtag encontrado será sustituido por la palabra 'HASHTAG'
re.sub(hashtag_pattern, hashtag_repl, url_tweet)  # Busca todos los hashtags en el tweet y los reemplaza por 'HASHTAG'
```
