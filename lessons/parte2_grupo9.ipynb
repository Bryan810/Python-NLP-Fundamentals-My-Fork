```python
# Importar librerías necesarias
import pandas as pd

# Cargar los datos desde el archivo CSV
csv_path = '../data/airline_tweets.csv'
tweets = pd.read_csv(csv_path, sep=',')

# Mostrar las primeras filas para inspeccionar los datos
tweets.head()
```

```python

second_example = tweets['text'][5]   # Selecciona el tweet en la fila con índice 5 (columna 'text')
second_example                       # Muestra el contenido del tweet (en Jupyter, la última variable se imprime automáticamente)
```

```python
second_example.strip()   # Elimina espacios en blanco (u otros caracteres como \n o \t) al inicio y al final del texto
```

```python
import re # Importacion modulo re para trabajar con expresiones regulares
```

```python
blankspace_pattern = r'\s+'    # Expresión regular: busca uno o más espacios en blanco (tab, espacio, salto de línea)
blankspace_repl = ' '          # Reemplazo: un solo espacio en blanco
```

```python
clean_text = re.sub(pattern = blankspace_pattern,   # Busca coincidencias de la expresión regular definida en blankspace_pattern
                    repl = blankspace_repl,         # Reemplaza cada coincidencia con un solo espacio
                    string = second_example)        # Texto de entrada donde se aplicará la sustitución

print(clean_text)  # Muestra el texto ya limpio, con espacios normalizados
```
