```python
from string import punctuation   # Importa la constante 'punctuation' del módulo 'string'
print(punctuation)               # Imprime todos los caracteres de puntuación definidos en ASCII
```

```python
def remove_punct(text):                               # Define una función que recibe un texto como entrada
    '''Remove punctuation marks in input text'''      
    
    no_punct = []                                     # Lista vacía para almacenar los caracteres válidos
    for char in text:                                 # Recorre cada carácter del texto
        if char not in punctuation:                   # Verifica si el carácter no es un signo de puntuación
            no_punct.append(char)                     # Si no lo es, lo añade a la lista
    text_no_punct = ''.join(no_punct)                 # Une la lista de caracteres en una cadena
    
    return text_no_punct                              # Devuelve el texto sin signos de puntuación
```

```python
import re
from string import punctuation

# Función para limpiar el texto
def clean_text_pipeline(text):
    '''Convierte a minúsculas, elimina puntuación y normaliza espacios'''
    
    # 1. Convertir a minúsculas
    text = text.lower()                              # Transforma todo el texto a minúsculas
    
    # 2. Eliminar signos de puntuación
    no_punct = []                                    # Lista para caracteres válidos
    for char in text:
        if char not in punctuation:                  # Solo se agregan los que no son signos de puntuación
            no_punct.append(char)
    text = ''.join(no_punct)                         # Une la lista en un string limpio
    
    # 3. Eliminar espacios en blanco adicionales
    blankspace_pattern = r'\s+'                      # Regex: busca 1 o más espacios/tabs/saltos de línea
    text = re.sub(blankspace_pattern, ' ', text)     # Reemplaza por un único espacio
    text = text.strip()                              # Quita espacios al inicio y final
    
    return text
```

```python
print("Texto limpio:", clean_text_pipeline(challenge1))
```
